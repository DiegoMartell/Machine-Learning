## 1. Fundamentos de la Técnica SLR

La **Regresión Lineal Simple (Simple Linear Regression - SLR)** es uno de los algoritmos de Machine Learning más básicos y fundamentales. Es un método estadístico que nos permite resumir y estudiar las relaciones entre dos variables continuas (cuantitativas):

* **Una variable es la predictora, explicativa o independiente ($X$).**
* **La otra es la variable de respuesta, dependiente o de salida ($Y$).**

**¿Cuál es el objetivo?**
El propósito de SLR es encontrar la "línea de mejor ajuste" (best-fitting straight line) a través de los datos que pueda usarse para predecir $Y$ a partir de $X$. Esta línea minimiza las distancias (errores o residuos) entre los puntos de datos reales y los puntos sobre la línea.

## 2. Modelo Matemático

Matemáticamente, la relación entre la variable independiente $X$ y la dependiente $Y$ se modela mediante la ecuación de una línea recta:

$$ Y = \beta_0 + \beta_1 X $$

Donde:
* **$Y$**: Es la variable que intentamos predecir (ej. Salario).
* **$X$**: Es la variable que estamos utilizando para hacer la predicción (ej. Años de Experiencia).
* **$\beta_0$ (Intersección o Sesgo - Bias)**: Es el valor esperado de $Y$ cuando $X = 0$. Gráficamente, es donde la línea cruza el eje Y.
* **$\beta_1$ (Pendiente - Slope o Peso)**: Es el cambio esperado en $Y$ por cada unidad de cambio en $X$. Determina la inclinación de la línea.

## 3. Implementación: Pipeline de Machine Learning

### Paso 1: Importar las librerías necesarias
Comenzamos importando las herramientas que usaremos.
* `pandas` para manipular los datos en formato de tabla.
* `matplotlib.pyplot` para crear los gráficos de dispersión y la línea de regresión.

```python
# Importamos pandas bajo el alias 'pd' para facilitar su uso
import pandas as pd
# Importamos pyplot de matplotlib bajo el alias 'plt' para generar visualizaciones
import matplotlib.pyplot as plt

print("✅ Librerías básicas cargadas.")
```

### Paso 2: Importar el dataset
Cargamos los datos y separamos nuestras variables.
* `X` será nuestra matriz de características (Años de Experiencia).
* `y` será nuestro vector de variables dependientes (Salario).

```python
# NOTA: Para este ejemplo, crearemos un dataset sintético pequeño
# que simule la relación entre Años de Experiencia (X) y Salario (y)
# para replicar la estructura del artículo original.
import numpy as np
np.random.seed(0)
experiencia = np.linspace(1, 10, 30)
salario = 30000 + (experiencia * 9000) + (np.random.randn(30) * 5000)
df = pd.DataFrame({'YearsExperience': experiencia, 'Salary': salario})

# Mostramos las primeras filas del dataframe
display(df.head())

# Extraemos la variable independiente (X). .iloc[:, :-1] selecciona todas las filas y todas las columnas menos la última.
# .values convierte el resultado en un array de NumPy bidimensional.
X = df.iloc[:, :-1].values

# Extraemos la variable dependiente (y). .iloc[:, 1] selecciona todas las filas y solo la segunda columna (índice 1).
y = df.iloc[:, 1].values
```

### Paso 3: Dividir el dataset en conjunto de Entrenamiento (Training) y Prueba (Test)
Es una buena práctica en Machine Learning dividir los datos. Entrenamos el modelo con una parte (ej. 80%) y evaluamos su rendimiento en la parte restante (20%) que el modelo nunca ha visto.
* Usaremos `train_test_split` de `sklearn.model_selection`.

```python
# Importamos la función train_test_split
from sklearn.model_selection import train_test_split

# Dividimos los datos (X e y).
# test_size = 1/3 significa que un 33% de los datos irá al set de prueba.
# random_state = 0 asegura que obtendremos la misma división aleatoria cada vez que ejecutemos el código.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=1/3, random_state=0)

print(f"Tamaño del set de entrenamiento: {X_train.shape[0]} muestras.")
print(f"Tamaño del set de prueba: {X_test.shape[0]} muestras.")
```

### Paso 4: Ajustar (Fitting) el modelo de Regresión Lineal Simple al set de Entrenamiento
Aquí es donde creamos el modelo matemático.
* Importamos la clase `LinearRegression` de `sklearn.linear_model`.
* Creamos un objeto (instancia) de esta clase llamado `regressor`.
* Usamos el método `.fit()` para que el modelo "aprenda" la relación entre `X_train` e `y_train` encontrando los valores óptimos para $\beta_0$ y $\beta_1$.

```python
# Importamos la clase LinearRegression
from sklearn.linear_model import LinearRegression

# Creamos una instancia del modelo
regressor = LinearRegression()

# Ajustamos el modelo a nuestros datos de entrenamiento
regressor.fit(X_train, y_train)

print("✅ Modelo entrenado.")
print(f"Intersección (Beta_0): {regressor.intercept_:.2f}")
print(f"Pendiente (Beta_1): {regressor.coef_[0]:.2f}")
```

### Paso 5: Predecir los resultados del set de Prueba
Ahora que el modelo está entrenado, le pedimos que prediga los salarios (`y_pred`) basándose en los años de experiencia del set de prueba (`X_test`).
* Usamos el método `.predict()`.

```python
# Generamos las predicciones utilizando los datos de entrada del set de prueba
y_pred = regressor.predict(X_test)

# Mostramos los resultados predichos frente a los reales
comparacion = pd.DataFrame({'Real': y_test, 'Predicción': y_pred})
display(comparacion)
```

### Paso 6: Visualizar los resultados del set de Entrenamiento
Graficaremos los puntos de datos reales de entrenamiento y la línea de regresión que nuestro modelo calculó.
* `plt.scatter` dibuja los puntos reales.
* `plt.plot` dibuja la línea de las predicciones.

```python
plt.figure(figsize=(8,5))
# Graficamos los datos reales de entrenamiento como puntos rojos
plt.scatter(X_train, y_train, color='red', label='Datos reales')

# Graficamos la línea de regresión.
# Eje X: Años de experiencia (X_train)
# Eje Y: Predicciones del modelo para X_train (regressor.predict(X_train))
# Color azul.
plt.plot(X_train, regressor.predict(X_train), color='blue', label='Línea de Regresión')

# Añadimos un título al gráfico
plt.title('Salario vs Experiencia (Set de Entrenamiento)')
# Etiquetamos el eje X
plt.xlabel('Años de Experiencia')
# Etiquetamos el eje Y
plt.ylabel('Salario')
plt.legend()
plt.grid(alpha=0.3)
# Mostramos el gráfico
plt.show()
```

### Paso 7: Visualizar los resultados del set de Prueba
Hacemos lo mismo, pero ahora trazando los puntos del set de prueba contra la misma línea de regresión calculada durante el entrenamiento. Esto nos permite ver visualmente qué tan bien el modelo generaliza a datos no vistos.

```python
plt.figure(figsize=(8,5))
# Graficamos los datos reales del set de prueba como puntos rojos
plt.scatter(X_test, y_test, color='red', label='Datos reales (Test)')

# Graficamos la MISMA línea de regresión.
# La línea no cambia, representa la ecuación matemática (Y = b0 + b1*X) aprendida en el paso 4.
plt.plot(X_train, regressor.predict(X_train), color='blue', label='Línea de Regresión (Modelo)')

# Añadimos título y etiquetas
plt.title('Salario vs Experiencia (Set de Prueba)')
plt.xlabel('Años de Experiencia')
plt.ylabel('Salario')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
